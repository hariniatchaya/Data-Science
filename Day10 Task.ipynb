```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("D:\\netflix_titles.csv")
print("First 5 rows of your dataset:")
print(df.head())
print("\nDataset Info:")
print(df.info())
```

```python
print("\nMissing values before treatment:")
missing_before = df.isnull().sum()
print(missing_before[missing_before > 0])
```

```python
duplicates = df.duplicated().sum()
print(f"Found {duplicates} duplicate rows")
```

```python
if duplicates > 0:
    df = df.drop_duplicates()
    print(f"✓ Removed {duplicates} duplicates")
    print(f"New dataset size: {df.shape}")
else:
    print("✓ No duplicates found")
```

```python
plt.figure(figsize=(4,4))
df['type'].value_counts().plot(kind='bar', color='skyblue')
plt.title('Bar Chart: type Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('barchart.png')
plt.show()
```

```python
df['type'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('type Distribution')
plt.ylabel('') 
plt.show()
```

```python
df['release_year'].plot(kind='hist', bins=50, edgecolor='black')
plt.title('release_year Distribution')
plt.xlabel('releases_year')
plt.ylabel('rating')
plt.show()
```
