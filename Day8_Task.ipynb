```python
import numpy as np

arr1 = np.array([10, 20, 30, 40, 50])
arr2 = np.linspace(0, 100, 11)
arr3 = np.random.randint(1, 100, size=(4, 4))

identity = np.eye(4)

print("1D Array:\n", arr1)
print("\nLinspace Array:\n", arr2)
print("\nRandom Matrix:\n", arr3)
print("\nIdentity Matrix:\n", identity)
```

```python
import numpy as np
matrix = np.random.randint(1, 100, (5,5))

print("Original Matrix:\n")
print(matrix)

print("\nElement at Row 2, Column 3:")
print(matrix[1,2])

print("\nSubmatrix:")
print(matrix[1:4,1:4])

print("\nValues Greater Than 50:")
print(matrix[matrix > 50])
```

```python
import numpy as np

data = np.random.randint(10, 100, 20)

print("Dataset:")
print(data)

print("\nMean:", np.mean(data))
print("Median:", np.median(data))
print("Standard Deviation:", np.std(data))
print("Variance:", np.var(data))
print("Maximum Value:", np.max(data))
print("Minimum Value:", np.min(data))
```

```python
import numpy as np
A = np.array([[2,4],[3,5]])
B = np.array([[1,2],[3,4]])

print("Matrix A:\n",A)
print("\nMatrix B:\n",B)

print("\nAddition:")
print(A + B)

print("\nMultiplication:")
print(np.dot(A,B))

print("\nDeterminant of A:")
print(np.linalg.det(A))

print("\nInverse of A:")
print(np.linalg.inv(A))

print("\nEigen Values:")
print(np.linalg.eig(A)[0])
```
